# MINMAX ALGO

```python
# WAP in Python to implement Minimax Algorithm
# Two Player Game

# Leaf node values
leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
# Minimax Algorithm using Recursion
def minimax(depth, node_index, is_maximizing):

    # Base condition:
    # If we reach a leaf node, return its value
    if depth == 3:
        return leaf_values[node_index]

    # MAX player
    if is_maximizing:

        left = minimax(
            depth + 1,
            node_index * 2,
            False
        )

        right = minimax(
            depth + 1,
            node_index * 2 + 1,
            False
        )

        return max(left, right)

    # MIN player
    else:

        left = minimax(
            depth + 1,
            node_index * 2,
            True
        )

        right = minimax(
            depth + 1,
            node_index * 2 + 1,
            True
        )

        return min(left, right)
        
# Main Program

print("MINIMAX ALGORITHM")
print("-----------------")

print("Leaf Node Values:")
print(leaf_values)

# MAX starts at depth 0 and node index 0
optimal_score = minimax(0, 0, True)

print("\nOptimal Score for MAX Player:", optimal_score)
```
